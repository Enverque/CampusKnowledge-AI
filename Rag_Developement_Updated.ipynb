```python
from pathlib import Path
import re

# Group all loaded pages by PDF
pdf_documents = {}

for doc in docs:
    source = doc.metadata.get("source", "")
    file_name = Path(source).name

    if file_name not in pdf_documents:
        pdf_documents[file_name] = []

    pdf_documents[file_name].append(doc)


print("======================================")
print("PDF GROUPING")
print("======================================")
print("Unique PDFs:", len(pdf_documents))
print("Total pages:", len(docs))
print("======================================")
```

```python
def detect_category(file_name, full_text):

    combined = (file_name + " " + full_text).lower()

    # Check specific categories first

    if "nirf" in combined or "national institutional ranking framework" in combined:
        return "NIRF Ranking"

    if any(x in combined for x in [
        "placement report",
        "placement brochure",
        "placement policy",
        "campus placement",
        "training and placement",
        "placements"
    ]):
        return "Placement"

    if any(x in combined for x in [
        "admission",
        "cap round",
        "institute level quota",
        "admission schedule"
    ]):
        return "Admission"

    if any(x in combined for x in [
        "fee structure",
        "tuition fee",
        "fees"
    ]):
        return "Fees"

    if any(x in combined for x in [
        "hostel",
        "hostel rules",
        "hostel regulations"
    ]):
        return "Hostel"

    if any(x in combined for x in [
        "student handbook",
        "student guide"
    ]):
        return "Student Handbook"

    if any(x in combined for x in [
        "attendance requirement",
        "attendance rules",
        "minimum attendance",
        "attendance"
    ]):
        return "Attendance"

    if any(x in combined for x in [
        "examination rules",
        "examination regulations",
        "exam rules",
        "academic regulations"
    ]):
        return "Examination"

    if any(x in combined for x in [
        "rules and regulations",
        "rules & regulations"
    ]):
        return "Rules & Regulations"

    return "Other"


def detect_year(file_name, full_text):

    combined = file_name + " " + full_text[:10000]

    # Academic year: 2025-26 / 2025–26
    academic_year = re.search(
        r"\b(20\d{2})\s*[-–]\s*(\d{2,4})\b",
        combined
    )

    if academic_year:
        return academic_year.group(0)

    # Otherwise find years
    years = re.findall(
        r"\b20\d{2}\b",
        combined
    )

    if years:
        # Count occurrences and use the most common year
        from collections import Counter

        year_counts = Counter(years)

        return year_counts.most_common(1)[0][0]

    return "Unknown"


def detect_document_type(file_name, category):

    name = file_name.lower()

    if "placement" in name and "report" in name:
        return "Placement Report"

    if "placement" in name and "brochure" in name:
        return "Placement Brochure"

    if "placement" in name and "policy" in name:
        return "Placement Policy"

    if "nirf" in name:
        return "NIRF Report"

    if "fee" in name:
        return "Fee Structure"

    if "handbook" in name:
        return "Student Handbook"

    if "hostel" in name:
        return "Hostel Rules"

    if "admission" in name or "cap" in name:
        return "Admission Document"

    if "rules" in name or "regulations" in name:
        return "Rules & Regulations"

    return category


print("Metadata functions ready!")
```

```python
from collections import Counter

pdf_metadata = {}

for file_name, pages in pdf_documents.items():

    # Combine text from the entire PDF
    full_text = "\n".join(
        page.page_content
        for page in pages
    )

    # Detect metadata ONCE for the entire PDF
    category = detect_category(
        file_name,
        full_text
    )

    year = detect_year(
        file_name,
        full_text
    )

    document_type = detect_document_type(
        file_name,
        category
    )

    # Store PDF-level metadata
    pdf_metadata[file_name] = {
        "category": category,
        "year": year,
        "document_type": document_type
    }


# Apply the same metadata to every page
for doc in docs:

    source = doc.metadata.get("source", "")
    file_name = Path(source).name

    metadata = pdf_metadata[file_name]

    doc.metadata["source_file"] = file_name
    doc.metadata["category"] = metadata["category"]
    doc.metadata["year"] = metadata["year"]
    doc.metadata["document_type"] = metadata["document_type"]


print("======================================")
print("PDF-LEVEL METADATA CREATED")
print("======================================")
print("PDFs processed:", len(pdf_metadata))
print("Pages updated:", len(docs))
print("======================================")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=120,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)

chunks = splitter.split_documents(docs)

print("======================================")
print("CHUNKING COMPLETE")
print("======================================")
print("Total pages:", len(docs))
print("Total chunks:", len(chunks))
print("======================================")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
test_embedding = embeddings.embed_query(
    "What is the placement information?"
)

print("======================================")
print("EMBEDDING TEST")
print("======================================")
print("Embedding length:", len(test_embedding))
print("First 10 values:", test_embedding[:10])
print("======================================")
```

```python
print("Finding problematic chunk...")

bad_indexes = []

for i, chunk in enumerate(chunks):
    try:
        embeddings.embed_documents([chunk.page_content])
    except Exception as e:
        bad_indexes.append((i, str(e)))
        print("\nProblem found!")
        print("Chunk index:", i)
        print("Source:", chunk.metadata.get("source_file"))
        print("Page:", chunk.metadata.get("page_number"))
        print("Category:", chunk.metadata.get("category"))
        print("Text preview:", repr(chunk.page_content[:300]))
        print("Error:", e)
        
        # Stop after finding the first one
        break

print("\n======================================")
print("CHECK COMPLETE")
print("======================================")
print("Problematic chunks found:", len(bad_indexes))
```

```python
print("Sanitizing chunk text...")

for chunk in chunks:
    if isinstance(chunk.page_content, str):
        # Convert unusual Unicode characters to safe text
        chunk.page_content = (
            chunk.page_content
            .encode("utf-8", errors="ignore")
            .decode("utf-8", errors="ignore")
        )

print("Sanitization complete.")
```

```python
print("Creating embeddings for all chunks...")
print("Total chunks:", len(chunks))

texts = [chunk.page_content for chunk in chunks]
metadatas = [chunk.metadata for chunk in chunks]

all_embeddings = embeddings.embed_documents(texts)

print("======================================")
print("ALL EMBEDDINGS CREATED")
print("======================================")
print("Number of embeddings:", len(all_embeddings))
print("Embedding dimension:", len(all_embeddings[0]))
```

```python
from langchain_community.vectorstores import FAISS

print("Creating FAISS vector database...")

vectorstore = FAISS.from_embeddings(
    text_embeddings=list(zip(texts, all_embeddings)),
    embedding=embeddings,
    metadatas=metadatas
)

print("======================================")
print("FAISS CREATED SUCCESSFULLY")
print("======================================")
print("Total vectors:", vectorstore.index.ntotal)
```

```python
query = "What is the placement information?"

results = vectorstore.similarity_search(query, k=5)

print("======================================")
print("SEARCH RESULTS")
print("======================================")

for i, doc in enumerate(results, 1):
    print(f"\n========== RESULT {i} ==========")
    print("Source:", doc.metadata.get("source_file"))
    print("Category:", doc.metadata.get("category"))
    print("Year:", doc.metadata.get("year"))
    print("Document Type:", doc.metadata.get("document_type"))
    print("Page:", doc.metadata.get("page_number"))
    print("\nText:")
    print(doc.page_content[:500])
```
